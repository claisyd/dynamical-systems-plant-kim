```python
import PyDSTool as dst
import numpy as np
from matplotlib import pyplot as plt
```

```python
DSargs = dst.args(name="Selkov")
pars = {"a": 0.1, "b": 0.6}
varspecs = {"x": "-x+a*y+x^2*y", "y": "b-a*y-x^2*y"}
ics = {"x": 0.75, "y": 1}
domain = {"x": [0, 5], "y": [0, 5]}

DSargs.pars = pars
DSargs.varspecs = varspecs
DSargs.ics = ics
DSargs.xdomain = domain
```

```python
from PyDSTool.Toolbox import phaseplane as pp

ode_sys = dst.Generator.Vode_ODEsystem(DSargs)
ode_sys.set(tdata=[0, 100])
print("Computing trajectory...")
trajectory = ode_sys.compute('dimensionless_run')
points = trajectory.sample()

# Time Series and Basic Trajectory
print("Plotting trajectory part A...")
fig_A1, ax_A1 = plt.subplots(figsize=(8, 8))
ax_A1.plot(points['t'], points['x'], label='x(t)', lw=2.5)
ax_A1.plot(points['t'], points['y'], label='y(t)', lw=2.5)
ax_A1.set_title('Time Series', fontsize=16)
ax_A1.set_xlabel('Time (dimensionless)', fontsize=12)
ax_A1.set_ylabel('Concentration (dimensionless)', fontsize=12)
plt.legend()
ax_A1.grid(True)
plt.show()

# Complete portrait
fp_coord = pp.find_fixedpoints(ode_sys, n=4, eps=1e-8)[0]
nulls_x, nulls_y = pp.find_nullclines(ode_sys, 'x', 'y', n=3, eps=1e-8, max_step=0.05, fps = [fp_coord])
fig_C, ax_C = plt.subplots(figsize=(8, 8))
ax_C.plot(points['x'], points['y'], 'k-', lw=2, label='Trajectory')
ax_C.plot(ics['x'], ics['y'], 'go', ms=10, label=f'Start ({ics["x"]},{ics["y"]})')
ax_C.set_title('Complete Phase Portrait', fontsize=16)
pp.plot_PP_vf(ode_sys, "x", "y", N=40, scale_exp=1)
plt.plot(nulls_x[:,0], nulls_x[:,1], 'r--', label='x-nullcline (dx/dt=0)')
plt.plot(nulls_y[:,0], nulls_y[:,1], 'b--', label='y-nullcline (dy/dt=0)')
ax_C.plot(fp_coord["x"], fp_coord["y"], 'm*', ms=15, label=f'Untable Steady State ({fp_coord["x"]:.2f},{fp_coord["y"]:.2f})')
ax_C.set_xlim(-0.1, 2.5)
ax_C.set_ylim(-0.1, 2.5) 
ax_C.legend()
ax_C.set_aspect('equal', adjustable='box')
plt.show()
```

```python
DSargs = dst.args(name = "Selkov")
pars = {"a": 0.1, "b": 0}
varspecs = {"x": "-x+a*y+x^2*y", "y": "b-a*y-x^2*y"}
ics = {"x": 0.6, "y": 1.3}
domain = {"x": [0, 5], "y": [0, 5]}

DSargs.pars = pars
DSargs.varspecs = varspecs
DSargs.ics = ics
DSargs.xdomain = domain
DSargs.tdata = [0, 100]

# Continuation Curve
ode = dst.Generator.Vode_ODEsystem(DSargs)
PC = dst.ContClass(ode)
PCargs = dst.args(name="EQ1", type="EP-C")
PCargs.freepars = ["b"]
PCargs.MaxNumPoints = 200
PCargs.MaxStepSize = 0.001
PCargs.MinStepSize = 1e-5
PCargs.LocBifPoints = "all"
PCargs.initpoint = fp_coord
PCargs.verbosity = 2
PC.newCurve(PCargs)

print("Computing continuation curve in parameter b...")
PC["EQ1"].forward()
PC["EQ1"].backward()

# Plot
PC.display(("b", "x"), stability=True)
```
